# 1. Compilación y ejecución


## Teoría

C es un lenguaje compilado. El flujo básico es:

`código fuente (.c) → compilador → ejecutable → ejecución`

### Compilar

```bash
gcc -Wall -Wextra -Wpedantic archivo.c -o programa
```

- `gcc`: ejecuta el compilador GCC.
- `-Wall`: activa muchas advertencias.
- `-Wextra`: activa advertencias adicionales.
- `-Wpedantic`: ayuda a detectar código que se aparta del estándar.
- `archivo.c`: código fuente.
- `-o programa`: nombre del ejecutable generado.

### Ejecutar

```bash
./programa
```

`./` indica que el ejecutable se encuentra en el directorio actual.

### Tipos de problemas

- **Error de compilación:** el ejecutable no se genera.
- **Warning:** el compilador detecta algo sospechoso; el programa puede compilar.
- **Error de ejecución:** el programa compiló, pero falla mientras corre.

### Seguridad

En C, que un programa compile no significa que sea correcto ni seguro. Los warnings deben revisarse.


````c
#include <stdio.h>

int main(void) {
    printf("Hola, mundo\n");
    return 0;
}
````

# 2. Variables y tipos de datos


## Teoría

Una variable es una zona de memoria utilizada para almacenar un valor.

Forma general:

```c
tipo nombre = valor;
```

| Tipo | Uso típico |
|---|---|
| `char` | carácter |
| `int` | entero |
| `float` | decimal |
| `double` | decimal con mayor precisión |

### Inicialización

```c
int numero = 10;
```

declara e inicializa.

```c
int numero;
```

solo declara. Una variable local sin inicializar no debe usarse como si ya contuviera un valor válido.

### `sizeof`

Permite consultar cuántos bytes ocupa un tipo o una variable.

### Seguridad

Usar datos no inicializados puede producir cálculos incorrectos y, más adelante, errores graves cuando esos datos se usan como índices, tamaños o direcciones.


````c
#include <stdio.h>

int main(void) {
    int edad = 20;
    char inicial = 'J';
    float altura = 1.75f;
    double nota = 4.35;

    printf("Edad: %d\n", edad);
    printf("Inicial: %c\n", inicial);
    printf("Altura: %.2f\n", altura);
    printf("Nota: %.2f\n", nota);

    printf("int: %zu bytes\n", sizeof(int));
    printf("char: %zu bytes\n", sizeof(char));
    printf("float: %zu bytes\n", sizeof(float));
    printf("double: %zu bytes\n", sizeof(double));

    return 0;
}
````

# 3. Operadores y conversiones


## Teoría

Operadores aritméticos:

```c
+  -  *  /  %
```

### División entera

`10 / 3` produce `3` si ambos operandos son enteros.

Si uno es decimal, por ejemplo `10 / 3.0`, el resultado es decimal.

### Cast

```c
(double)a
```

convierte temporalmente `a` a `double` para esa expresión.

### Truncamiento

```c
double x = 8.75;
int y = x;
```

`y` queda con `8`.

### Seguridad

Errores de tipos, truncamiento y overflow pueden producir tamaños de memoria incorrectos y conducir a accesos fuera de límites.


````c
#include <stdio.h>

int main(void) {
    int a = 10;
    int b = 3;

    printf("Suma: %d\n", a + b);
    printf("Resta: %d\n", a - b);
    printf("Multiplicacion: %d\n", a * b);
    printf("Division entera: %d\n", a / b);
    printf("Residuo: %d\n", a % b);
    printf("Division decimal: %.2f\n", (double)a / b);

    return 0;
}
````

# 4. Comparaciones y condicionales


## Teoría

```c
==  !=  >  <  >=  <=
&&  ||  !
```

En C:

- `0` se interpreta como falso.
- cualquier valor distinto de `0` se interpreta como verdadero.

### `if`, `else if`, `else`

Las condiciones se evalúan de arriba hacia abajo.

### Error clásico

```c
x = 5;   // asignación
x == 5;  // comparación
```

### Seguridad

Una condición incorrecta puede afectar controles de acceso, validaciones y decisiones de autorización.


````c
#include <stdio.h>

int main(void) {
    int edad = 18;
    int tienePermiso = 1;

    if (edad >= 18 && tienePermiso) {
        printf("Acceso permitido\n");
    } else {
        printf("Acceso denegado\n");
    }

    return 0;
}
````

# 5. switch


## Teoría

`switch` permite seleccionar entre varios valores concretos.

### `break`

`break` termina la ejecución del `switch`.

Si se omite, C continúa con los casos siguientes: **fall-through**.

### Seguridad

Un `break` olvidado puede provocar la ejecución accidental de acciones adicionales.


````c
#include <stdio.h>

int main(void) {
    int opcion = 2;

    switch (opcion) {
        case 1:
            printf("Consultar archivo\n");
            break;
        case 2:
            printf("Modificar archivo\n");
            break;
        case 3:
            printf("Eliminar archivo\n");
            break;
        default:
            printf("Opcion no valida\n");
    }

    return 0;
}
````

# 6. Ciclos


## Teoría

### `while`

Evalúa la condición antes de cada repetición.

### `for`

Agrupa inicialización, condición y actualización.

### Ciclo infinito

Si la variable de control no cambia, la condición puede permanecer verdadera indefinidamente.

### Seguridad

Un ciclo mal controlado puede consumir CPU, bloquear un proceso o provocar accesos fuera de límites cuando recorre memoria.


````c
#include <stdio.h>

int main(void) {
    int contador = 1;

    while (contador <= 10) {
        printf("%d\n", contador);
        contador++;
    }

    return 0;
}
````

# 7. Memoria y direcciones


## Teoría

Una variable tiene:

1. un **valor**;
2. una **dirección de memoria**.

Si:

```c
int i = 5;
```

entonces:

- `i` representa el valor `5`.
- `&i` representa la dirección de memoria de `i`.

Esta diferencia entre **valor** y **dirección** es la base para entender apuntadores.

### Seguridad

C permite trabajar directamente con direcciones de memoria. Eso da control, pero también hace posibles accesos inválidos y corrupción de memoria.


````c
#include <stdio.h>

int main(void) {
    int numero = 10;

    printf("Valor: %d\n", numero);
    printf("Direccion: %p\n", (void *)&numero);

    return 0;
}
````

# 8. Apuntadores


## Teoría

Un apuntador es una variable cuyo valor es una **dirección de memoria**.

```c
int *p;
```

### `&`

Obtiene la dirección de una variable.

### `*`

Desreferencia el apuntador: accede al valor almacenado en la dirección que contiene.

```c
int numero = 10;
int *p = &numero;
```

Entonces:

```text
numero   → 10
&numero  → dirección de numero
p        → dirección de numero
*p       → 10
```

### Modificación indirecta

```c
*p = 50;
```

modifica la memoria a la que `p` apunta.

### Seguridad

Nunca desreferenciar un apuntador que no apunte a una dirección válida.


````c
#include <stdio.h>

int main(void) {
    int numero = 10;
    int *p = &numero;

    printf("numero: %d\n", numero);
    printf("*p: %d\n", *p);
    printf("&numero: %p\n", (void *)&numero);
    printf("p: %p\n", (void *)p);

    *p = 50;

    printf("numero despues: %d\n", numero);

    return 0;
}
````

# 9. Apuntadores a apuntadores


## Teoría

Un apuntador también es una variable, por lo tanto también tiene una dirección.

```c
int numero = 10;
int *p = &numero;
int **pp = &p;
```

```text
pp → p → numero
```

- `p` contiene la dirección de `numero`.
- `pp` contiene la dirección de `p`.
- `*pp` obtiene el contenido de `p`.
- `**pp` llega al valor de `numero`.

Cada `*` sigue un nivel de indirección.


````c
#include <stdio.h>

int main(void) {
    int numero = 10;
    int *p = &numero;
    int **pp = &p;

    printf("numero: %d\n", numero);
    printf("*p: %d\n", *p);
    printf("**pp: %d\n", **pp);

    **pp = 99;
    printf("numero despues: %d\n", numero);

    return 0;
}
````

# 10. Arreglos y aritmética de apuntadores


## Teoría

Un arreglo almacena elementos consecutivos en memoria.

```c
int v[5] = {10, 20, 30, 40, 50};
```

En muchas expresiones, `v` se comporta como la dirección de su primer elemento:

```c
v == &v[0]
```

### Aritmética de apuntadores

Si `p` es `int *`, `p + 1` avanza un `int`, no un byte.

### Equivalencias clave

```c
v       == &v[0]
v + i   == &v[i]
*(v+i)  == v[i]
p[i]    == *(p+i)
```

### Seguridad

C no comprueba automáticamente que un índice esté dentro del arreglo. Un acceso fuera de límites puede leer o sobrescribir memoria ajena.


````c
#include <stdio.h>

int main(void) {
    int numeros[5] = {10, 20, 30, 40, 50};
    int *p = numeros;

    printf("*p: %d\n", *p);
    printf("*(p + 1): %d\n", *(p + 1));
    printf("p[2]: %d\n", p[2]);

    printf("p: %p\n", (void *)p);
    printf("p + 1: %p\n", (void *)(p + 1));

    return 0;
}
````
