### Switch 
Sirve cuando tenemos una variable que puede tomar varios valores concretos. La guía lo presenta como una alternativa especializada a una serie de if, especialmente para valores enteros o caracteres.


````python
switch (opcion) {
    case 1:
        printf("Crear usuario\n");
        break;

    case 2:
        printf("Eliminar usuario\n");
        break;

    case 3:
        printf("Listar usuarios\n");
        break;

    default:
        printf("Opcion invalida\n");
}
````

El break lo que hace es saltar de un ciclo a otro.
El while es un cuclo que evalúa la condición antes de cada repetición,debido a eso puede ejecutarse 0 veces si la condición es falsa desde un inicio

````python
#Ejemplo 
#Include <stdio.h>
int main(void){
    int contador = 1; 
    while(contador <= 5){
        printf("%d\n, contador);
        contador++;
    }
    return 0; 
}
````
